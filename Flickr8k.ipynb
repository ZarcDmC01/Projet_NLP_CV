```python
import os
from os import listdir
from pickle import dump
from keras.applications.vgg16 import VGG16
from keras.preprocessing.image import load_img
from keras.preprocessing.image import img_to_array
from keras.applications.vgg16 import preprocess_input
from keras.models import Model

# extract features from each photo in the directory
def extract_features(directory):
    # load the model
    model = VGG16()
    # re-structure the model (remove the last classification layer)
    model = Model(inputs=model.inputs, outputs=model.layers[-2].output)
    # summarize
    print(model.summary())
    
    # extract features from each photo
    features = dict()
    for name in listdir(directory):
        # Create the full path to the file
        filename = os.path.join(directory, name)
        
        # 🛡️ Security: Skip if it's a directory instead of a file
        if os.path.isdir(filename):
            print(f"> Ignored directory: {name}")
            continue
            
        # 🛡️ Security: Only process common image formats
        if not name.lower().endswith(('.png', '.jpg', '.jpeg')):
            continue
            
        # load an image from file
        image = load_img(filename, target_size=(224, 224))
        # convert the image pixels to a numpy array
        image = img_to_array(image)
        # reshape data for the model
        image = image.reshape((1, image.shape[0], image.shape[1], image.shape[2]))
        # prepare the image for the VGG model
        image = preprocess_input(image)
        # get features
        feature = model.predict(image, verbose=0)
        # get image id (remove the file extension)
        image_id = name.split('.')[0]
        # store feature
        features[image_id] = feature
        print('>%s' % name)
        
    return features

# define the correct directory path containing your images
# (Adjust this path if your images are inside another sub-folder)
directory = r'Data/Flickr8k_Dataset'

# extract features from all images
features = extract_features(directory)
print('Extracted Features: %d' % len(features))

# save to file
with open('features.pkl', 'wb') as f:
    dump(features, f)
```

### Nettoyage du texte

```python


def load_doc(filename):
    """Charge le fichier texte brut."""
    with open(filename, 'r', encoding='utf-8') as file:
        text = file.read()
    return text

def load_descriptions(doc):
    """
    Organise le texte brut en un dictionnaire.
    Clé : nom de l'image (ex: '1000268201_693b08cb0e.jpg')
    Valeur : liste des 5 légendes associées.
    """
    mapping = {}
    # Découpage ligne par ligne
    for line in doc.split('\n'):
        # Ignorer les lignes vides
        if len(line) < 2:
            continue
        
        # Séparer l'identifiant de l'image et la légende
        tokens = line.split('\t')
        image_id, image_desc = tokens[0], tokens[1:]
        
        # Retirer l'indice de la légende (ex: '1000268201_693b08cb0e.jpg#0' -> '1000268201_693b08cb0e.jpg')
        image_id = image_id.split('#')[0]
        
        # Reconstituer la description en texte si elle a été découpée
        image_desc = ' '.join(image_desc)
        
        # Initialiser ou ajouter à la liste du dictionnaire
        if image_id not in mapping:
            mapping[image_id] = []
        mapping[image_id].append(image_desc)
        
    return mapping

def clean_descriptions(descriptions):
    """
    Nettoie les descriptions selon le cahier des charges :
    - Passage en minuscules
    - Suppression de la ponctuation
    - Suppression des mots contenant des chiffres
    - Ajout des jetons de début (startseq) et de fin (endseq)
    """
    # Création d'une table de traduction pour supprimer la ponctuation efficacement
    table = str.maketrans('', '', string.punctuation)
    
    for key, desc_list in descriptions.items():
        for i in range(len(desc_list)):
            desc = desc_list[i]
            
            # 1. Séparation en mots (Tokenisation basique par espace)
            words = desc.split()
            
            # 2. Conversion en minuscules 
            words = [word.lower() for word in words]
            
            # 3. Suppression de la ponctuation de chaque mot 
            words = [word.translate(table) for word in words]
            
            # 4. Suppression des jetons vides et des mots contenant des chiffres 
            words = [word for word in words if len(word) > 0 and word.isalpha()]
            
            # 5. Reconstruction de la phrase nettoyée
            cleaned_desc = ' '.join(words)
            
            # 6. Ajout des jetons impératifs de début et de fin [cite: 73]
            desc_list[i] = f"startseq {cleaned_desc} endseq"

# =====================================================================
# EXÉCUTION DU SCRIPT
# =====================================================================

# 1. Définition du chemin d'accès vers votre fichier de tokens
filename = r'Data\Flickr8k_text\Flickr8k.token.txt'

if not os.path.exists(filename):
    raise FileNotFoundError(f"Le fichier {filename} est introuvable. Vérifiez l'emplacement.")

# 2. Lancement de la chaîne de traitement
raw_doc = load_doc(filename)
descriptions = load_descriptions(raw_doc)

print(f"[INFO] Nombre d'images chargées dans le dictionnaire : {len(descriptions)}")

# 3. Nettoyage
clean_descriptions(descriptions)

# 4. Vérification du résultat sur une image exemple
example_key = list(descriptions.keys())[0]
print(f"\n[EXEMPLE] Légendes nettoyées pour l'image '{example_key}' :")
for caption in descriptions[example_key]:
    print(f"  -> {caption}")
```

```python
import os
import numpy as np
import pickle
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.utils import to_categorical

# =====================================================================
# 1. FONCTION POUR CHARGER LES LES ID D'IMAGES (TRAIN / TEST)
# =====================================================================
def load_set(filename):
    """Charge une liste d'identifiants d'images à partir d'un fichier (ex: Flickr_8k.trainImages.txt)"""
    with open(filename, 'r', encoding='utf-8') as file:
        text = file.read()
    
    dataset = list()
    for line in text.split('\n'):
        # Ignorer les lignes vides
        if len(line) < 1:
            continue
        # On récupère l'identifiant (nom de l'image)
        dataset.append(line)
    return set(dataset)

# Charger les lignes d'entraînement officielles
train_file = r'Data\Flickr8k_text\Flickr_8k.trainImages.txt'
train_images = load_set(train_file)
print(f"[INFO] Nombre d'images pour l'entraînement : {len(train_images)}")

# =====================================================================
# 2. CHARGEMENT DES FICHIERS PRÉCÉDENTS (FEATURES & TOKENIZER)
# =====================================================================
# Charger les caractéristiques VGG16
with open(os.path.join("Data", "vgg16_features.pkl"), "rb") as f:
    all_features = pickle.load(f)

# Charger le tokenizer
with open(os.path.join(r"Data", "tokenzier.pkl"), "rb") as f:
    tokenizer = pickle.load(f)

# Récupérer les variables globales calculées à l'étape d'avant
vocab_size = len(tokenizer.word_index) + 1
max_length = 34 # Mettez ici la valeur exacte affichée par votre script précédent (souvent 34 ou 35)

# =====================================================================
# 3. LE GÉNÉRATEUR DE DONNÉES (DATA GENERATOR)
# =====================================================================
def data_generator(descriptions, features, tokenizer, max_length, vocab_size, batch_size=32):
    """
    Générateur de données qui fournit des lots (batches) de données au modèle.
    Retourne : [Entrée_Image, Entrée_Texte], Sortie_Mot_Suivant
    """
    X1, X2, y = list(), list(), list()
    n = 0
    
    # Boucle infinie demandée par Keras pour les générateurs d'entraînement
    while True:
        for key, desc_list in descriptions.items():
            # On vérifie si l'image fait partie de notre set d'entraînement
            if key not in train_images:
                continue
                
            n += 1
            # Récupérer la feature de l'image (et enlever la dimension de batch inutile pour l'instant)
            image_feature = features[key][0]
            
            for desc in desc_list:
                # Transformer la phrase en liste de chiffres (Tokens)
                seq = tokenizer.texts_to_sequences([desc])[0]
                
                # Découper la séquence en paires d'entrées/sorties mot par mot
                for i in range(1, len(seq)):
                    # Séparer en entrée (mots vus jusqu'ici) et sortie (mot suivant à deviner)
                    in_seq, out_seq = seq[:i], seq[i]
                    
                    # Remplir de zéros (Padding) l'entrée pour atteindre 'max_length'
                    in_seq = pad_sequences([in_seq], maxlen=max_length)[0]
                    
                    # Convertir le mot de sortie en "One-Hot Encoding" (vecteur de la taille du vocabulaire)
                    out_seq = to_categorical([out_seq], num_classes=vocab_size)[0]
                    
                    # Stocker dans nos listes de batch
                    X1.append(image_feature)
                    X2.append(in_seq)
                    y.append(out_seq)
            
            # Dès que nous avons accumulé assez d'exemples pour faire un "batch"
            if n == batch_size:
                yield [np.array(X1), np.array(X2)], np.array(y)
                # Réinitialisation des listes pour le prochain batch
                X1, X2, y = list(), list(), list()
                n = 0

print("[INFO] Générateur de données prêt à l'emploi !")
```

### Architecture du Modèle (CNN + RNN)

```python
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input,Dense,LSTM,Embedding,Dropout,add
from tensorflow.keras.utils import plot_model
import os
```

```python
# Branche1: l'encodeur d'iamge (CNN features)

inputs1=Input(shape=(4096,),name="Entree_image")
fe1=Dropout(0.5) (inputs1)
fe2=Dense(256,activation='relu')(fe1)


# Branche 2: le décodeur de texte (séquence de mots)

inputs2= Input(shape=(max_length,),name="etntree texte")

se1 = Embedding(input_dim=vocab_size, output_dim=256, mask_zero=True)(inputs2)

# couche récurente (LSTM) pour capter la dependance temporelle des mots

se2=Dropout(0.5)(se1)
se3=LSTM(256)(se2)

decoder1=add([fe2,se3])
decoder2=Dense(256,activation='relu')(decoder1)

outputs=Dense(vocab_size,activation='softmax')(decoder2)
```
